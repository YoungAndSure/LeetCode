```python
from typing import List
class Solution:
    def findMedianSortedArrays(self, nums1: List[int], nums2: List[int]) -> float:
        nums = sorted(nums1 + nums2)
        if len(nums) % 2 == 1:
            return nums[len(nums)//2]
        return (nums[len(nums)//2]+nums[len(nums)//2-1])/2
```

```python
s = Solution()
print(s.findMedianSortedArrays([1,3], [2]))
print(s.findMedianSortedArrays([1,2], [3,4]))
print(s.findMedianSortedArrays([0,0], [0,0]))
print(s.findMedianSortedArrays([], [1]))
print(s.findMedianSortedArrays([2], []))
print(s.findMedianSortedArrays([-10,-9,-8], [1,2]))
```

```python
from typing import List
class Solution2:
    def findMedianSortedArrays(self, nums1: List[int], nums2: List[int]) -> float:
        MIN = -10**6
        MAX = 10**6
        if len(nums1) > len(nums2) :
            nums1, nums2 = nums2, nums1

        m = len(nums1)
        n = len(nums2)
        # i的含义是，nums1里分割线左边的元素个数，
        # left和right是i的左右边界。
        # nums1里分割线左边的元素个数，可以是0个，也可以是整个数组也就是m个,
        # 所以left = 0, right = m
        left = 0
        right = m
        while left <= right :
            i = (left + right) // 2
            # 左侧的元素个数应该是全体元素个数的一半。所以(i+j) = (m+n+1)//2
            # 如果总数是奇数个，因为加了1，所以左侧就包含了中位数，且中位数是左侧所有元素里最大的那个
            j = (m + n + 1) // 2 - i
            #print(i, j)
            # i可以取到0，所以i-1有可能小于0
            maxLeft1 = nums1[i-1] if i > 0 else MIN
            # i可以取到m
            minRight1 = nums1[i] if i < m else MAX
            # i = 0 时，j = (m + n + 1) // 2, j > 0
            # j - n = (m + n + 1) // 2 - n = (m - n + 1) // 2
            # m - n <= 0, 所以j - n <=0, 所以 j <= n
            # 结论就是j有可能等于n，但j - 1肯定小于n

            # i = m 时，j = (n - m + 1)//2, j > 0
            # j - 1 = (n - m + 1)//2 - 1 = (n - m - 1) // 2, 有可能等于0
            # 因此，j不可能等于0，但j-1可能小于0，需要防范，所以：
            maxLeft2 = nums2[j-1] if j-1 >= 0 else MIN
            minRight2 = nums2[j] if j < n else MAX
                
            #print("maxLeft1:{}, minRight1:{}, maxLeft2:{}, minRight2:{}".format(maxLeft1, minRight1, maxLeft2, minRight2))
            
            if (maxLeft1 <= minRight2 and maxLeft2 <= minRight1) :
                if (m + n) % 2 == 0 :
                    return (max(maxLeft1, maxLeft2) + min(minRight1, minRight2))/2
                else :
                    return max(maxLeft1, maxLeft2)
            if maxLeft1 > minRight2 :
                right = i - 1
            elif minRight1 < maxLeft2 :
                left = i + 1
        return 0.0

s = Solution2()
print(s.findMedianSortedArrays([1,3], [2]))
print(s.findMedianSortedArrays([1,2], [3,4]))
print(s.findMedianSortedArrays([0,0], [0,0]))
print(s.findMedianSortedArrays([], [1]))
print(s.findMedianSortedArrays([2], []))
print(s.findMedianSortedArrays([-10,-9,-8], [1,2]))
```

#### 难点
- i和j的含义到底是什么，确定它们的取值范围
- i的左右边界是什么
- i,i-1,j,j-1可能越界，但根据公式关系，可以确定越界的范围
