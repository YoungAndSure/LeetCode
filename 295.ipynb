中位数跟数据的分布有关，如果数据一直在一个范围内波动，中位数就也在这个范围内波动。如果数据在逐渐增大，中位数也逐渐增大。

```python
from typing import *
import heapq

class MedianFinder:

    def __init__(self):
        self.left=[]
        self.right=[]
        self.index = 0

    def addNum(self, num: int) -> None:
        MAX = 10**6
        MIN = -10**6
        maxLeft = self.left[-1] if len(self.left) > 0 else MAX
        minRight = self.right[0] if len(self.right) > 0 else MIN
        if num < maxLeft :
            heapq.heappush_max(self.left, (num, self.index))
            self.index += 1
            if len(self.left) > (len(self.left) + len(self.right) + 1) // 2 :
                heapq.heappush(self.right, heapq.heappop_max(self.left))
        elif num > minRight :
            heapq.heappush(self.right, (num, self.index))
            self.index += 1
            if len(self.right) > (len(self.left) + len(self.right)) // 2 :
                heapq.heappush_max(self.left, heapq.heappop(self.right))
        
    def findMedian(self) -> float:
        if (len(self.left) + len(self.right)) % 2 == 0:
            return (self.left[-1] + self.right[0])/2
        else :
            return self.left[-1]
        
obj = MedianFinder()
obj.addNum(num)
param_2 = obj.findMedian()
```
